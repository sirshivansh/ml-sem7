```python
df = pd.read_csv('Iris_3.csv')
df.head()
```

```python
df.columns
```

```python
df['species'].unique()
```

```python
df1 = df[df['species'] == "setosa"]
df2 = df[df['species'] == "versicolor"]
df3 = df[df['species'] == "virginica"]
```

```python
plt.scatter(df1['petal_length'], df1['petal_width'], c='red')
plt.scatter(df2['petal_length'], df2['petal_width'], c='blue')
plt.scatter(df3['petal_length'], df3['petal_width'], c='green')

plt.xlabel("petal length")
plt.ylabel("petal width")

plt.show()
```

```python
X = df.drop('species', axis='columns')
Y = df.species
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y)
```

```python
from sklearn.svm import SVC

svm = SVC(kernel='linear')
```

```python
svm.fit(X_train, Y_train)
predicted = svm.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(Y_test, predicted)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(Y_test, predicted)

sns.heatmap(cm, annot=True, cmap="Blues")
plt.show()
```
